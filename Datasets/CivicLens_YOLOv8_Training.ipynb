```python
# Step 3: Train YOLOv8 Master Model
from ultralytics import YOLO
import glob

yaml_files = glob.glob('**/data.yaml', recursive=True)
yaml_path = yaml_files[0] if yaml_files else 'civic-issue-detection-project-jc7ii-1/data.yaml'
print(f"Using master dataset config: {yaml_path}")

model = YOLO('yolov8n.pt')
results = model.train(
    data=yaml_path,
    epochs=40,
    imgsz=640,
    batch=16,
    name='civiclens_master_model'
)
```

```python
# Step 4: Export Master Model Weights to ONNX (Backend) & TFLite (Android App)
model.export(format='onnx')
model.export(format='tflite')
print("🏆 Master Model Training Complete! Model weights saved inside runs/detect/civiclens_master_model/weights/")
```
